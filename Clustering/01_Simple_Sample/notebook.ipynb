```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
from sklearn.metrics import (
    calinski_harabasz_score,
    davies_bouldin_score,
    silhouette_score,
)
```

```python
pd.set_option("display.max_colwidth", None)
pd.set_option("display.width", None)
pd.set_option("display.max_columns", None)
```

```python
X, _ = make_blobs(n_samples=300, centers=4, cluster_std=1.0, random_state=42)
_
```

```python
features = ["Feature_1", "Feature_2"]
label = "Cluster"
df = pd.DataFrame(X, columns=features)
df.head()
```

```python
plt.scatter(df["Feature_1"], df["Feature_2"])
plt.xlabel("Feature_1")
plt.ylabel("Feature_2")
```

```python
model = KMeans(n_clusters=3, random_state=42)
labels = model.fit_predict(df)
df[label] = labels
df.head()
```

```python
plt.scatter(df["Feature_1"], df["Feature_2"], c=df["Cluster"])
plt.xlabel("Feature_1")
plt.ylabel("Feature_2")
plt.tight_layout()
```

```python
X = df[features]
lables = df[label]

silhouette = silhouette_score(X=X, labels=labels)
davies_bouldin = davies_bouldin_score(X=X, labels=labels)
calinski_harabasz = calinski_harabasz_score(X=X, labels=labels)
metrics = {
    "Silhouette": silhouette,
    "Davies Bouldin": davies_bouldin,
    "Calinski Harabasz": calinski_harabasz,
}
pd.DataFrame(metrics.items(), columns=["Metric", "Value"])
```
