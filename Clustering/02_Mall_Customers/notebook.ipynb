```python
import pandas as pd
from modelens import ClusteringAnalyzer, clustering_models
```

```python
raw_data = pd.read_excel("dataset.xlsx")
df = raw_data.copy()
```

```python
if "CustomerID" in df.columns:
    df = df.drop(columns=["CustomerID"])

df = df.rename(
    columns={
        "Annual Income (k$)": "Annual Income",
        "Spending Score (1-100)": "Spending Score",
        "Education ": "Education",
    }
)
df["Gender"] = df["Gender"].map({"M": 1, "F": 0})
```

```python
X = df[
    [
        "Gender",
        "Age",
        "Annual Income",
        "Spending Score",
        "Marital Status",
        "Education",
    ]
].copy()

X = pd.get_dummies(
    X,
    columns=["Marital Status", "Education"],
    dtype=int,
)

base_features = [
    "Gender",
    "Age",
    "Annual Income",
    "Spending Score",
]

education_features = [col for col in X.columns if col.startswith("Education_")]

marital_features = [col for col in X.columns if col.startswith("Marital Status_")]

feature_sets = {
    "Base": base_features,
    "Base + Education": base_features + education_features,
    "Base + Marital Status": base_features + marital_features,
    "Base + Education + Marital Status": base_features
    + education_features
    + marital_features,
}
```

```python
analyzer=ClusteringAnalyzer(df=X)
```
